```python
# Partiendo del csv data_basic.csv, con los arreglos básicos ya realizados (one-hot, eliminación de id y variable redundante, etc.)

import pandas as pd
import numpy as np

SEED = 777
np.random.seed(SEED)
```

```python
# si por lo que sea se ejecuta esto desde un windows hay que tener cuidado con el formato de la ruta 
ruta_csv = '../data/scaledic.csv'

data_basic = pd.read_csv(ruta_csv) #csv con preprocesado básico sin normalizar
```

```python
from sklearn.model_selection import train_test_split

# separación de características y objetivo
X_basic = data_basic.drop('Variable de Salida', axis=1)
y_basic = data_basic['Variable de Salida']

# separación en train y test
X_basic_train, X_basic_test, y_basic_train, y_basic_test = train_test_split(X_basic, y_basic, test_size=0.2, random_state = SEED, shuffle=True, stratify=y_basic)
```

```python
# en esta celda voy a crear un csv por cada tipo de estandarización/normalización realizado

from sklearn.preprocessing import MaxAbsScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import RobustScaler
from sklearn.preprocessing import StandardScaler

# incialización de los escaladores
escaladores = {
    'max_abs': MaxAbsScaler(),
    'min_max': MinMaxScaler(),
    'robust': RobustScaler(),     # si hay outlyers este puede ser muy útil
    'standard': StandardScaler()
}


# bucle para hacer todos las combinaciones
for nombre, scaler in escaladores.items():
    # se fitea y transforma con el train y solo se transforma con el test para evitar data leak
    X_train_transformed = scaler.fit_transform(X_basic_train)
    X_test_transformed = scaler.transform(X_basic_test)

    # se convierte de nuevo a data frame de pandas
    X_train_transformed = pd.DataFrame(X_train_transformed, columns=X_basic_train.columns, index=X_basic_train.index)
    X_test_transformed = pd.DataFrame(X_test_transformed, columns=X_basic_test.columns, index=X_basic_test.index)

    # añadimos de nuevo el target para tener el csv entero
    data_train_transformed = pd.concat([X_train_transformed, y_basic_train], axis=1)
    data_test_transformed = pd.concat([X_test_transformed, y_basic_test], axis=1)

    # esto es para guardar el csv en la correspondiente carpeta, lo voy a dejar comentado,
    # pero si alguien lo necesita por lo que sea que lo descomente.

     
"""     
    data_train_transformed.to_csv(f'../data/scaled/data_train{nombre}.csv', index = False)
    data_test_transformed.to_csv(f'../data/scaledta_test{nombre}.csv', index = False)
 """
    
    






```
