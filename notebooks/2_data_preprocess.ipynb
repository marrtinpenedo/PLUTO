Partiendo de los resultados del primer análisis (`1_raw_analysis.ipynb`) creamos el conjunto de datos con las transformaciones básicas y necesarias del que partirán las demás pruebas: eliminación de variables redundantes (Variable 02), tomamos como nulo el valor 0 que tomaba la variable categórica Variable 92, hacemos One-Hot a las variables categóricas y transformamos la variable de salida a un valor entero: `NOK = 0; OK = 1`.

Gracias al primer análisis sabemos que no hay valores nulos ni duplicados.

```python
# Importaciones necesarias y semilla
import pandas as pd
```

```python
# Cargamos los datos 
data = pd.read_excel(io='../data/raw/Dataset_01_Anonimizado.xlsx', decimal=',')
```

```python
# Cambiamos valor de la variable de salida
data['Variable de Salida'] = data['Variable de Salida'].apply(lambda x: 1 if x == 'OK' else 0)

# Eliminamos la variable Variable 02
data.drop(columns=['Variable 02'], inplace=True)

# One-Hot Encoding 
categorical_columns = [i for i in data.columns if data[i].dtypes not in ['int64', 'float64']]
data_encoded = pd.get_dummies(columns=categorical_columns, data=data)
```

```python
# Guardamos los datasets "básicos"
data_encoded.to_csv('../data/processed/data_basic.csv', index=False)
```
