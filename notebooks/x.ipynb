```python
import pandas as pd
```

```python
# Cargamos los datos 
Train = pd.read_csv('../data/scaled/data_trainstandard.csv')
Test = pd.read_csv('../data/scaled/data_teststandard.csv')

# El split se realiza en 3_creacion_csvs_escalados.ipynb
# Estratificado y 80/20

X_train = Train.drop(columns=['Variable de Salida'])
y_train = Train['Variable de Salida']

X_test = Test.drop(columns=['Variable de Salida'])
y_test = Test['Variable de Salida']
```

```python
# Hacemos K vecinos y vemos el clasification rport
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, f1_score, make_scorer, make_scorer, make_scorer

# Hacemos grid search para encontrar el mejor número de vecinos, y demás parámetros
from sklearn.model_selection import GridSearchCV

# Definimos el modelo
knn = KNeighborsClassifier()
# Definimos los parámetros a probar
param_grid = {
    'n_neighbors': [1, 2, 3, 5, 7, 9, 15, 30, 50],
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan', 'minkowski','l1', 'l2', 'cosine', 'cityblock', 'nan_euclidean'],
    'algorithm': ['auto', 'ball_tree', 'kd_tree', 'brute'],
    'leaf_size': [20, 30, 40, 50],

}
# Hacemos el grid search y le damos la vuelta al f1 socre ( que lo haga para la clase 0)
grid_search = GridSearchCV(estimator=knn, param_grid=param_grid, cv=5, n_jobs=6,
                           scoring=make_scorer(f1_score, pos_label=0), verbose=2)
grid_search.fit(X_train, y_train)

print("Mejores parámetros: ", grid_search.best_params_)

# Obtenemos el mejor modelo
best_knn = grid_search.best_estimator_


# Hacemos predicciones
y_pred = best_knn.predict(X_test)

# Mostramos el classification report
print(classification_report(y_test, y_pred))
```

```python
from imblearn.ensemble import BalancedRandomForestClassifier

# creamos un gird search basico 

# Definimos el modelo
brf = BalancedRandomForestClassifier(random_state=42)
# Definimos los parámetros a probar
param_grid_brf = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['auto', 'sqrt', 'log2']
}
# Hacemos el grid search y le damos la vuelta al f1 socre ( que lo haga para la clase 0)
grid_search_brf = GridSearchCV(estimator=brf, param_grid=param_grid_brf, cv=5, n_jobs=6,
                               scoring=make_scorer(f1_score, pos_label=0), verbose=2)
grid_search_brf.fit(X_train, y_train)   

print("Mejores parámetros: ", grid_search_brf.best_params_)
# Obtenemos el mejor modelo
best_brf = grid_search_brf.best_estimator_
# Hacemos predicciones
y_pred_brf = best_brf.predict(X_test)
# Mostramos el classification report
print(classification_report(y_test, y_pred_brf))
```
