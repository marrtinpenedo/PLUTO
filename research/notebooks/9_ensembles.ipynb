## Ensembles

Se probarán diferentes ensembles. 

```python
# Importaciones necesarias y semilla 
import pandas as pd
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, HistGradientBoostingClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.feature_selection import RFECV
from sklearn.metrics import f1_score, classification_report

SEED = 777
```

```python
# Cargamos los datos 
Train = pd.read_csv('../data/scaled/data_trainstandard.csv')
Test = pd.read_csv('../data/scaled/data_teststandard.csv')

# El split se realiza en 3_creacion_csvs_escalados.ipynb
# Estratificado y 80/20

X_train = Train.drop(columns=['Variable de Salida'])
y_train = Train['Variable de Salida']

X_test = Test.drop(columns=['Variable de Salida'])
y_test = Test['Variable de Salida']
```

### Random Forest

#### Sin feature selection

##### Con CV

```python
est = RandomForestClassifier(random_state=SEED)
cv = StratifiedKFold(shuffle=True, random_state=SEED)

param_grid = {
    'n_estimators': [30, 50, 100, 150, 200, 300, 500],
    'criterion': ['gini', 'entropy', 'log_loss'],
    'bootstrap': [True, False],
    'class_weight': ['balanced', 'balanced_subsample']
}

grid = GridSearchCV(
    estimator=est,
    param_grid=param_grid,
    cv=cv,
    scoring='f1',
    n_jobs=6
    
)

grid.fit(X_train, y_train)

best_model = grid.best_estimator_
```

```python
y_pred = best_model.predict(X_test)
f1_score(y_test, y_pred)
```

##### Sin CV

```python
est = RandomForestClassifier(random_state=SEED)

param_grid = {
    'n_estimators': [5, 10, 30, 50, 100, 150, 200, 300, 500],
    'criterion': ['gini', 'entropy', 'log_loss'],
    'bootstrap': [True, False],
    'class_weight': ['balanced', 'balanced_subsample']
}

grid = GridSearchCV(
    estimator=est,
    param_grid=param_grid,
    scoring='f1',
    n_jobs=6  
)

grid.fit(X_train, y_train)

best_model = grid.best_estimator_
```

```python
grid.best_score_
```

```python
y_pred = best_model.predict(X_test)
f1_score(y_test, y_pred)
```

##### Con CV

```python
# TODO

est = HistGradientBoostingClassifier(random_state=SEED)
cv = StratifiedKFold(shuffle=True, random_state=SEED)

param_grid = {
    'learning_rate': [0.001, 0.01, 0.1, 0.5, 0.7, 1, 2, 5],
    'class_weight': [{0:3, 1:1}, {0:5, 1:1}, {0:10, 1:1}, {0:15, 1:1}],
    'interaction_cst': ['pairwise', 'no_interactions', None],
    'max_bins': [255, 500]
}
 # Se supone que con variables continuas que tienen mucha resolución 
 # y donde importa los bins está bien subirlo

grid = GridSearchCV(
    estimator=est,
    param_grid=param_grid,
    cv=cv,
    scoring='f1_weighted',
    n_jobs=6
    
)

grid.fit(X_train, y_train)

best_model = grid.best_estimator_
```

```python
y_pred = best_model.predict(X_test)
print(classification_report(y_test, y_pred))
print(grid.best_params_)
print(f1_score(y_test, y_pred))
```
